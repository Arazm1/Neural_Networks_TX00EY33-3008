## Loading the dataset and preparing

```python
from ucimlrepo import fetch_ucirepo
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
# Fetch Connectionist Bench (Sonar, Mines vs. Rocks) by its UCI ID (151)
#sonar = fetch_ucirepo(id=151)
#df = sonar.data.original

import pandas as pd

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/undocumented/connectionist-bench/sonar/sonar.all-data"
df = pd.read_csv(url, header=None)
```

```python
df.sample(5)
```

```python
df.shape
```

 Check if any of the columns have null values

```python
df.isna().sum()
```

There are no **null values.**

Check the column names

```python
df.columns
```

Currently the column names are just numeric

```python
df[60].value_counts()
```

Column 60 has **111 samples of metal cylinder** and **97 samples of rocks**

Divide the dataset into input features `X` and target features `y`.

```python
X = df.drop(60, axis='columns')
y = df[60]
```

Convert the **column 60** 'M' and 'R' into 1.0 and 0.0.

```python
y = (df[60] == 'M').astype('float32')
y.sample(5)
```

```python
y.value_counts()
```

Now split the dataset into **train data** and **test data.** 

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=1
)
```

```python
X_train.shape, X_test.shape
```

```python
import tensorflow as tf
from tensorflow import keras
```

```python
model = keras.Sequential([
    keras.layers.Dense(60, input_dim=60, activation='relu'),
    keras.layers.Dense(30, activation='relu'),
    keras.layers.Dense(15, activation='relu'),
    keras.layers.Dense(1, activation='sigmoid'),

])

model.compile(
    loss='binary_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

model.fit(
    X_train,
    y_train,
    epochs=100,
    batch_size=8)
```

Lets see how the Model performs with test data

```python
model.evaluate(X_test, y_test)
```

The model predicted correctly only 75% correct

```python
y_pred = model.predict(X_test).reshape(-1)
print(y_pred[:10])

# Round the values to nearest integer ie 0 or 1
y_pred = np.round(y_pred)
print(y_pred[:10])
```

```python
from sklearn.metrics import confusion_matrix, classification_report

print(classification_report(y_test, y_pred))
```

```python
model = keras.Sequential([
    keras.layers.Dense(60, input_dim=60, activation='relu'),
    keras.layers.Dropout(0.5),
    keras.layers.Dense(30, activation='relu'),
    keras.layers.Dropout(0.5),
    keras.layers.Dense(15, activation='relu'),
    keras.layers.Dropout(0.5),
    keras.layers.Dense(1, activation='sigmoid'),

])

model.compile(
    loss='binary_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

model.fit(
    X_train,
    y_train,
    epochs=100,
    batch_size=8)
```

```python
model.evaluate(X_test, y_test)
```

```python
y_pred = model.predict(X_test).reshape(-1)

# Round the values to nearest integer ie 0 or 1
y_pred = np.round(y_pred)

print(classification_report(y_test, y_pred))
```
